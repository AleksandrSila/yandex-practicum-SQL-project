## Передем к запросам

### Посчитайте, сколько книг вышло после 1 января 2000 года

```python
SQL("""
       SELECT COUNT(book_id) AS cnt_books
       FROM books
       WHERE publication_date > '2000-01-01';
    """)
```

После 2000 года бышло 819 книг

### Для каждой книги посчитайте количество обзоров и среднюю оценку

```python
SQL('''SELECT books.title, 
              books.book_id AS id,
              COUNT(DISTINCT(review_id)) as total_reviews, 
              AVG(ratings.rating) as avg_rating
                     
        FROM books
        LEFT JOIN reviews ON books.book_id = reviews.book_id
        LEFT JOIN ratings ON books.book_id = ratings.book_id
        GROUP BY books.book_id
        ORDER BY total_reviews DESC;
          ''')
```

Мы получили таблицу, которая содержит название, id, количество обзоров и среднюю оценку

### Определите издательство, которое выпустило наибольшее число книг толще 50 страниц — так вы исключите из анализа брошюры

```python
SQL('''SELECT publishers.publisher,
                     COUNT(books.book_id) as books_total
              FROM books
              LEFT JOIN publishers ON books.publisher_id=publishers.publisher_id
              WHERE books.num_pages >= 50
              GROUP BY publishers.publisher
              ORDER BY books_total DESC
              LIMIT 1;
          ''')
```

Издательство Penguin Books выпустило наибольшее количество книг больше 50 страниц, их общее количество равняется 42

### Определите автора с самой высокой средней оценкой книг — учитывайте только книги с 50 и более оценками

```python
SQL('''SELECT author, 
             AVG(temp.avg_rating) AS average_rating,
             COUNT(temp.book_id) AS books_total
             FROM  (SELECT author, 
                           books.book_id, 
                           AVG(rating) AS avg_rating
                    FROM authors
                    LEFT JOIN books ON books.author_id = authors.author_id
                    LEFT JOIN ratings ON ratings.book_id = books.book_id
                    GROUP BY author, books.book_id
                    HAVING COUNT(rating) >= 50) AS temp
            GROUP BY author
            ORDER BY average_rating DESC
            LIMIT 1
           
           
        ''')
```

Самым высокооцениваемым автором оказалась J.K. Rowling, со средним рейтингом 4.283844 и общим количеством книг в размере 4

### Посчитайте среднее количество обзоров от пользователей, которые поставили больше 48 оценок

```python
SQL('''SELECT ROUND(AVG(total_reviews))
              FROM  (
                     SELECT reviews.username,
                            COUNT(reviews.review_id) AS total_reviews
                     FROM reviews
                     WHERE reviews.username IN  (
                                                 SELECT ratings.username
                                                 FROM ratings
                                                 GROUP BY ratings.username
                                                 HAVING COUNT(ratings.rating) > 48
                                                )
                     GROUP BY reviews.username
                    ) AS reviews_count;
            ''')
```

Среднее количество обзоров от пользователей, которые поставили больше 48 оценок - 24 обзора
